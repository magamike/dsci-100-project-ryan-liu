Visualization 1: Relationship of Conf Win and Regional Win.

This visualization serves as a confirmation for the team's overall regional and conference performance. As expected, the correlation is positive because a higher win in either tend to indicate a more skillful team. This is also an indication for the overall spread of bid types. Because classification requires two predictors with distinct seperation, this graph tells us that the conf win pct and the reg win pct are somewhat mediocre but still very grouped categories. 

```python
wba_visual1 = alt.Chart(women_basketball_drop).mark_point(opacity=0.4).encode(
    alt.X("reg_wins_pct").title("Regional Win Percentage").scale(zero=False),
    alt.Y("conf_wins_pct").title("Conference Win Percentage").scale(zero=False),
    alt.Color("bid").title("Bid Type")
)
wba_visual1
```

Visualization 2: Ratio of bid types at Different Win Pct (Conference)

This is supplementary to the first visualization and acts as a way to fully confirm the validity of the bid type distributions. Auto bids are granted to teams that win the conference while at-large bids are granted to teams with strong performance. The ratio of auto to at-large bids is surpassed at higher win percentage.
```python
wba_visual2 = alt.Chart(women_basketball_drop).mark_bar().encode(
    alt.X("conf_wins_pct").title("Conference Win Percentage"),
    alt.Y("count()").title("School Count"),
    alt.Color("bid").title("Bid Type")
)
wba_visual2
```
